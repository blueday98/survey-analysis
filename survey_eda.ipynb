```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import font_manager
```

```python
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", 80)

# 현재 환경에 설치된 한글 폰트를 직접 등록
font_path = "/usr/share/fonts/truetype/wqy/wqy-zenhei.ttc"
font_manager.fontManager.addfont(font_path)
korean_font = font_manager.FontProperties(fname=font_path).get_name()

plt.style.use("seaborn-v0_8-whitegrid")
plt.rcParams.update({
    "font.family": korean_font,
    "axes.unicode_minus": False,
    "figure.facecolor": "#FCFBF8",
    "axes.facecolor": "#FCFBF8",
    "grid.color": "#E7E3DC",
    "grid.alpha": 0.7,
    "axes.titleweight": "bold",
    "axes.titlesize": 14,
})

COLORS = ["#5278A3", "#5B9A8B", "#D97863", "#71AFA3", "#D9A441"]
```

```python
df = pd.read_csv("data/raw/러닝 자세 분석 서비스 사용자 조사 — 실사용본(응답) - 설문지 응답 시트1.csv")

# 타임스탬프와 후속 인터뷰 이름은 개인정보이므로 즉시 제외
df = df.drop(columns=[df.columns[0], df.columns[-1]])
```

```python
print("응답자 수:", len(df))
print("분석 문항 수:", len(df.columns))
print("데이터 크기:", df.shape)
```

```python
missing_summary = df.isna().sum().sort_values(ascending=False)
missing_summary[missing_summary > 0].to_frame("결측치 수")
```

## 러닝 빈도

```python
running_counts = df.iloc[:, 0].value_counts().sort_values()
ax = running_counts.plot.barh(figsize=(8, 4), color=COLORS[0], width=0.65)
ax.set_title("최근 6개월 러닝 빈도")
ax.set_xlabel("응답 인원")
ax.set_ylabel("")
ax.bar_label(ax.containers[0], padding=3)
plt.tight_layout()
plt.show()

runner_count = int((df.iloc[:, 0] != "전혀 하지 않았다").sum())
print(f"최근 6개월 러닝 경험자: {runner_count}/{len(df)}명 ({runner_count/len(df):.1%})")
```

최근 6개월 동안 한 번이라도 러닝한 응답자가 다수를 차지합니다.

## 핵심 의향 지표

```python
score_columns = {
    "도움 기대": df.columns[10],
    "사용 의향": df.columns[11],
    "촬영 의향": df.columns[12],
    "업로드 부담": df.columns[13],
    "개선 적용 의향": df.columns[21],
    "재촬영 의향": df.columns[22],
    "지불 의향": df.columns[23],
}
high_scores = pd.Series({name: int((df[column] >= 4).sum()) for name, column in score_columns.items()}).sort_values()

ax = high_scores.plot.barh(figsize=(8, 5), color=COLORS[1], width=0.65)
ax.set_title("핵심 지표 4~5점 응답")
ax.set_xlabel("응답 인원 (전체 21명)")
ax.set_ylabel("")
ax.bar_label(ax.containers[0], labels=[f"{v}/21" for v in high_scores], padding=3)
plt.tight_layout()
plt.show()

for name, count in high_scores.sort_values(ascending=False).items():
    print(f"{name}: {count}/{len(df)}명 ({count/len(df):.1%})")
```

서비스 사용과 개선 적용 의향은 높지만, 촬영과 지불 단계에서는 비율이 낮아집니다.

## MVP 우선 기능

```python
mvp_column = df.columns[18]
mvp_options = [
    "측정값과 피드백이 나온 이유 및 판단 기준 설명",
    "촬영 위치와 방법을 단계별로 안내",
    "이전 영상과 새로운 영상의 변화 비교",
    "영상이 분석 가능한지 자동 확인",
]
mvp_counts = pd.Series({
    option: int(df[mvp_column].fillna("").str.contains(option, regex=False).sum())
    for option in mvp_options
}).sort_values()

ax = mvp_counts.plot.barh(figsize=(9, 4), color=COLORS[4], width=0.65)
ax.set_title("처음 출시할 때 필요한 기능")
ax.set_xlabel("선택 인원 (전체 21명)")
ax.set_ylabel("")
ax.bar_label(ax.containers[0], labels=[f"{v}/21" for v in mvp_counts], padding=3)
plt.tight_layout()
plt.show()
```
